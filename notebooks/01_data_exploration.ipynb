# 01 - Data exploration
Run `python run_pipeline.py` first, then run these cells from the `notebooks/` folder.

```python
import pandas as pd, json
pd.set_option('display.width',200); pd.set_option('display.max_colwidth',120)
df = pd.read_csv('../outputs/events_correlated.csv.gz', low_memory=False, dtype=str).fillna('')
print(df.shape)
df.event_type.value_counts().head(15)
```

```python
# the analyst timeline (notable events only)
tl = pd.read_csv('../outputs/timeline.csv')
tl[['timestamp','host','user','kind','summary']].head(40)
```

```python
# process chains that matter
chains = json.load(open('../outputs/process_chains.json'))
for c in chains:
    if c['notable']:
        print(c['chain_id'], c['host'], c['start'][11:19], c['n_events'], 'events\n   ', c['narrative'][:200])
```

```python
# alerts sorted by risk
al = pd.DataFrame(json.load(open('../outputs/alerts.json')))
al.sort_values('risk_score', ascending=False)[['alert_id','rule_id','host','user','first_seen','risk_score','severity','title']].head(15)
```

```python
# the incident timeline
inc = json.load(open('../outputs/incidents.json'))[0]
print(inc['title'], inc['risk_score'])
print('\n'.join(inc['timeline']))
```

```python
# ML view: most anomalous process starts
ml = pd.read_csv('../outputs/ml_scores.csv').merge(df[['event_uid','host','process','parent_process','command_line']], on='event_uid')
ml.sort_values('ml_score', ascending=False).head(15)
```
